```python

import requests
import re
from pprint import pprint
import json
req = requests.get('https://www.youtube.com/watch?v=WifF8ncArM4')
# linkdata = re.findall(r'"description":{"runs":(.*?)},"subscribeButton"', req.text)
linkdata = re.findall(r'"commandMetadata":{"webCommandMetadata":{"url":(.*?),"webPageType":"WEB_PAGE_TYPE_CHANNEL"', req.text)
# '"commandMetadata":{"webCommandMetadata":{"url":"/channel/UCmbs8T6MWqUHP1tIQvSgKrg","webPageType":"WEB_PAGE_TYPE_CHANNEL"'
pprint(linkdata)
links = []
max = 20
flag = 0
for i in linkdata:
    if i.startswith('"/channel/'):
        links.append(i.replace('"',''))
    flag += 1
    if flag >= max:
        break
links = list(set(links))
print(links)
    
```
