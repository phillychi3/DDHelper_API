```python
import requests
import re
from fake_useragent import UserAgent
ua = UserAgent()
```

```python
LINK="https://www.twitch.tv/mobilmobil"
user_agent = ua.random
r = requests.get(LINK, headers={'User-Agent': user_agent})
print(re.search(r'isLiveBroadcast',r.content.decode('utf-8')))
```
